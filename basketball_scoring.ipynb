```python
import pandas as pd

# Load dataset
df = pd.read_csv('/content/stephen_curry_shots_2023.csv')

df.head()
```

```python
# Overall Shooting Statistics

# Probabilities of make / miss
prob_make = df['result'].mean()
prob_miss = 1 - prob_make

# Proportions of shot types (three-pointers and two-pointers)
prop_three = (df['shot_type'] == 3).mean()
prop_two = (df['shot_type'] == 2).mean()

# Probability of lead
prob_lead = df['lead'].mean()

print(f"Overall probability of a make (result = TRUE): {prob_make:.4f}")
print(f"Overall probability of a miss (result = FALSE): {prob_miss:.4f}")
print(f"Overall proportion of three-pointers (shot_type = 3): {prop_three:.4f}")
print(f"Overall proportion of two-pointers (shot_type = 2): {prop_two:.4f}")
print(f"Overall probability of having the lead (lead = True): {prob_lead:.4f}")
```

```python
# Probabilities

prob_3_of_4_shots = pow(prob_make, 3) * prob_miss * 4
prob_4_of_5_3pointers = pow(prop_three, 4) * prop_two * 5

print(f"Probability of Steph making 3 of the next 4 shots: {prob_3_of_4_shots:.4}")
print(f"Probability that 4 of the next 5 shots are three-pointers: {prob_4_of_5_3pointers:.4}")
```

```python
# Conditional Probabilities - Future

# If the next shot is a three-pointer... probability he makes it:
prob_three_given_make = df[df['result'] == True]['shot_type'].eq(3).mean()
bayes_make_given_three = (prob_three_given_make * prob_make) / prop_three

# Probability it was taken while his team had the lead (lead = True):
prob_three_given_lead = df[df['lead'] == True]['shot_type'].eq(3).mean()
bayes_lead_given_three = (prob_three_given_lead * prob_lead) / prop_three

# Now the two-pointers ... same logic
prob_two_given_make = df[df['result'] == True]['shot_type'].eq(2).mean()
bayes_make_given_two = (prob_two_given_make * prob_make) / prop_two

# Probability it was taken while his team had the lead (lead = True):
prob_two_given_lead = df[df['lead'] == True]['shot_type'].eq(2).mean()
bayes_lead_given_two = (prob_two_given_lead * prob_lead) / prop_two

print("If the next shot Steph shoots is a three-pointer...")
print(f"  What is the probability he makes it? {bayes_make_given_three:.4f}")
print(f"  What is the probability his team had the lead? {bayes_lead_given_three:.4f}")
print()
print("If the next shot Steph shoots is a two-pointer...")
print(f"  What is the probability he makes it? {bayes_make_given_two:.4f}")
print(f"  What is the probability his team had the lead? {bayes_lead_given_two:.4f}")
```

```python
# Conditional Probabilities - Past

# If Steph just made a shot... probability that it was a three-pointer:

# P(Make | 3-Pointer)
prob_make_given_three = df[df['shot_type'] == 3]['result'].mean()
# P(3-Pointer)
prop_three = (df['shot_type'] == 3).mean()

# Bayes: P(3-Pointer | Make) = [P(Make | 3-Pointer) * P(3-Pointer)] / P(Make)
bayes_three_given_make = (prob_make_given_three * prop_three) / prob_make

# Now for the two-pointer:

# P(Make | 2-Pointer)
prob_make_given_two = df[df['shot_type'] == 2]['result'].mean()
# P(2-Pointer)
prop_two = (df['shot_type'] == 2).mean()

# Bayes: P(2-Pointer | Make) = [P(Make | 2-Pointer) * P(2-Pointer)] / P(Make)
bayes_two_given_make = (prob_make_given_two * prop_two) / prob_make

print("If Steph just made a shot...")
print(f"  What is the probability it was a three-pointer? {bayes_three_given_make:.4f}")
print(f"  What is the probability it was a two-pointer? {bayes_two_given_make:.4f}")
```
